```python
class Environment:

    def __init__(self, state = 'Dirty'):
        self.state = state

    def get_percept(self):
        return self.state
    
    def clean_room(self):
        self.state = 'Clean'

#this utility based asgent works in a little of the similar sort of fashion on how a learning agent work based agent works
#only in the sense that it can be manaegd via a dictionary to store a key : value pair for 
# key = state and value = score (something along the same lines of reward/punishment) 

# one distinct thinf to keep in mind a utility based agent has to track the number of utilities it has for itself

# and based on the percept this agent will select an acction that best suits itself just like a learning based agent 
# also selects an action before acting upon that action  

class UtilityBasedAgent:

    def __init__(self):
        self.utility = {'Dirty': -10, 'Clean': 10}

    def calculate_utility(self, percept):
        return self.utility[percept] 

    def select_action(self, percept):
        if percept == 'Dirty':
            return 'Clean the room'
        else:
            return 'No action needed'

    def act(self, percept):
        return self.select_action(percept)


def run_agent(environment, agent, steps):

    total_utility = 0
    
    for step in range(steps):
        percept = environment.get_percept()
        action = agent.act(percept)

        utility = agent.calculate_utility(percept)
        total_utility += utility

        if action == 'Clean the room':
            environment.clean_room()
        
        print(f'Steps: {step + 1} -> Percept: {percept} -> Action: {action} -> Utility: {utility}')

    print(f"Total Utility: {total_utility}")

agent = UtilityBasedAgent()
environment = Environment()

run_agent(environment, agent, 5)
```
