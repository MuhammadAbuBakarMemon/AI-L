```python
class Environment:

    def __init__(self, state = 'Dirty'):
        self.state = state

    def get_percept(self):
        return self.state
    
    def clean_room(self):
        self.state = 'Clean'

class GoalBasedAgent:

    def __init__(self, goal='Clean'):
        self.goal = goal

    #formulate goal shapes the due action for us 
    #bascially we change our goal to achieve on the basis of the percept our agent gets from it's sensors 
    def formulate_goal(self, percept):
        if percept == 'Dirty':
            self.goal = 'Clean'
        else:
            self.goal = 'No action needed'

    def act(self, percept):

        self.formulate_goal(percept)

        if self.goal == 'Clean':
            return 'Clean the room'
        else:
            return 'No action needed room is clean'
       

def run_agent(environment, agent, steps):

    for step in range(steps):
        percept = environment.get_percept()
        action = agent.act(percept)

        #if percept == 'Dirty':
        #logically not sound - RECTIFICATION: The environment changes based on the agent's ACTION
        if action == 'Clean the room':    
            environment.clean_room()

        print(f'Steps: {step + 1} -> Percept: {percept} -> Actions: {action}')

agent = GoalBasedAgent()
environment = Environment()

run_agent(environment, agent, 5)
```
