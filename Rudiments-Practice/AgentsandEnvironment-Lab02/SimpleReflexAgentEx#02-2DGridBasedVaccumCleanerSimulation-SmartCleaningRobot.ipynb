```python
#This method is how it (the question was solved in the Lab Manual) 

class SimpleReflexAgent:

    def __init__(self):
        self.position = 0
        # tracks agent position in the grid

    #pass the grid so that the function works on the actual grid 
    def act(self, percept, grid):

        if percept == 'Dirty':
            #grid[self.position] = 'Clean' not required idk why was it in the code 
            #this also violates how AI Agents and Environments are supposed to Interact in AI Simulations
            #IN AI DESIGN AN AGENT SHOULD NEVER DIRECTLY MANIPULATE THE ENVIRONMENT'S INTERNAL VARIABLE,
            return "Clean the room"

        elif percept == 'Clean': 
            return "Room is already clean"
        else: 
            pass

    def move(self):

        if self.position < 8:
            self.position += 1
        
        return self.position
```

```python
def run_agent(environment, agent, steps):

    for step in range (steps):

        percept = environment.get_percept(agent.position)
        action = agent.act(percept, environment.grid)

        if percept == 'Dirty':
            environment.clean_room(agent.position)
            
        print(f'Step: {step + 1} -> Position: {agent.position} -> Percept: {percept} -> Action: {action}')
        environment.display_grid(agent.position)
        agent.move()

agent = SimpleReflexAgent()
environment = Environment()

run_agent(environment, agent, 9)
```
