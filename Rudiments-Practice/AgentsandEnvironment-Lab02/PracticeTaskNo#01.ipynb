Design a reflex-based cleaning agent that perceives the cleanliness of a room. If the
room is dirty, the agent cleans it. If the room is already clean, it does nothing except
confirming the cleanliness. The agent does not have memory or predictive capabilities
and reacts solely to the current percept.

A. Modify the code so that the initial state of the environment (&#39;Dirty&#39; or &#39;Clean&#39;) is
randomly set when the program starts. The state should be chosen at the
beginning of the program and should vary each time it runs.

B. Modify the code to make the environment’s state change randomly after each
step. The state should switch between &#39;Dirty&#39; and &#39;Clean&#39; after each iteration of
the loop or after the agent performs an action. This will simulate a dynamic
environment where the state is unpredictable throughout the agent’s execution.

```python
import random 

class Environment:

    def __init__(self):
        self.state = random.choice(['Dirty', 'Clean'])

        #random - library 
        #choice - function that takes any sequence and returns one random element from it

    def get_percept(self):
        return self.state
    
    def clean_room(self):
        self.state = 'Clean'

    def randomize_state(self):
        self.state = random.choice(['Dirty', 'Clean'])

class SimpleReflexAgent:

    def __init__(self):
        pass

    def act(self, percept):
        if percept == 'Dirty':
            return 'Clean the room'
        else:
            return 'Room is already clean'
        

#the simulation function simulates interactions between the agent and the environment for a specified number of steps 

def run_agent(agent, environment, steps):

    for step in range(steps):
        percept = environment.get_percept()
        action = agent.act(percept)

        print(f'Percept: {percept}, Action: {action}')

        if percept == 'Dirty':
            environment.clean_room()

        environment.randomize_state()

#creating instance for agent and anvironment 
agent = SimpleReflexAgent()
environment = Environment()

#run the agent in the environment for 5 steps 

run_agent(agent, environment, 5)
```
