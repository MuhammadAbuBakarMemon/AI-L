In this task, the agent chooses a movie to watch by considering multiple factors, such as
reviews and mood. A utility-based agent uses a utility function to evaluate different
options and select the one that maximizes overall satisfaction. This agent balances
between positive reviews and the current mood to make an optimal decision.
Environment Class: Environment
Represents the environment that provides a set of movies and their respective reviews.
movies: A dictionary where keys are movie titles and values are review scores
(e.g., ratings from 1 to 10).
get_percept(): Returns the list of movies and their review scores.

```python
class Environment:

    def __init__(self, movie):
        self.movie = movie 
        #we provide a list of movies in a dictionary format 
        # key : value pair - Movie Title - Score 

    def get_percept(self):
        return self.movie
```

Agent Class: UtilityBasedAgent
Decides which movie to watch by using a utility function that balances the review score
and a mood factor.
mood_factor: Represents how much mood influences the choice (a float
between 0 and 1). Higher values indicate stronger mood influence.
utility(movie, review): Computes utility by combining review scores and mood.
For simplicity, we assume:
○ Utility = (review score) * (mood factor)
act(percept): Chooses the movie with the highest utility value.

```python
class UtilityBasedAgent:

    def __init__(self, mood_factor):
        self.mood_factor = mood_factor

    def calculate_utility(self, review):
        return review * self.mood_factor
    
    #or we can name it act as well 
    def select_movie(self, movies):

        best_movie = None
        best_utility = - float('inf')   #used usually when we want to find the maximum value possible 

        for movie, review in movies.items():
            current_utility = self.calculate_utility(review)

            if current_utility > best_utility:
                best_movie = movie
                best_utility = current_utility

        return best_movie
```

Simulation Function: run_agent(agent, environment)
Simulates a single decision by the agent based on the available movies and their
reviews.
Steps: Since this is a one-time decision, there’s no loop required.

```python
def run_agent(environment, agent):

    percept = environment.get_percept()
    best_choice = agent.select_movie(percept)

    print(f'Avavilable Movies: {percept}')
    print(f'Best Movie: {best_choice}')
```

```python
Movie = {'Movie A' : 30, 'Movie B' : 7, 'Movie C' : 12}

agent = UtilityBasedAgent(mood_factor = 0.8)
environment = Environment(Movie)

run_agent(environment, agent)
```
