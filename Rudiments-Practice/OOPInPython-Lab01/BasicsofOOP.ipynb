```python
'''
OOP allows one to develop program that is easier to maintain and extend 

Key concepts/pillars of OOP include
1- Encapsulation: hiding internal details 
2- Inheritance: resusing code from existing classes 
3- Polymorphism: using the same interface for different objects 
4- Abstraction: focusing on essential features while hiding complexity 

'''
```

```python
'''
A Class is a blueprint or template for creating objects. It defines the attributes (data) and the methods (function) an 
object created from the class will have.

An object is an instance of a class. 
It represents a real entity created using the class's blueprint with it's own values for attributes

'''
```

```python
#Creating a class in python
class Student:
    pass

#Creating an Object in Python
s1 = Student()
s2 = Student()
```

```python
'''
Attributes in pyhton are the variables that belong to a class and hold data or properties of an object 
Methods are the functions defined inside a class that define the behvior or actions of an object 

Attributes store information specific to an object while Methods allow objects to perform operations using that data 

Using attributes and methods together help in organizing code in a structured and reuseable way, making programs easier to 
understand and maintain 
'''
```

```python
#Sample class and calling it's object and method

class MyClass:
    x = 5 #x is an attribute 

    def method_one(self):
        print("This is method one....")

m1 = MyClass()
print(m1.x)
m1.method_one()
```
