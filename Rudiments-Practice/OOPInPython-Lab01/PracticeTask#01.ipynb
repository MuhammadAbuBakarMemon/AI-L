```python
class SmartLight:

    def __init__(self, room_name):
        self.room_name = room_name 
        self.status = 'OFF'

    def switch_ON(self):
        self.status = 'ON'
        print(f'{self.room_name} light tuned ON.')

    def switch_OFF(self):
        self.status = 'OFF'
        print(f'{self.room_name} light tuned OFF.')

    def display_status(self):
        print(f'{self.room_name} light is {self.status}.')

l1 = SmartLight('Bedroom')
l2 = SmartLight('Living Room')

l1.switch_OFF()
l2.switch_ON()

l2.display_status()
l1.display_status()

```
