```python
'''
A Constructor is a special method in a class that automatically called when an object is cerated. It is used to initialize the objects of a class.
In Python the constructor method is always named __init__()

'''
```

```python
class Student:
    
    #this is a class attribute (shared by all instances) no need of that static keyword here
    school = 'Nixor'
    def __init__(self, name, age):
        self.name = name #instance attribute 
        self.age = age #instance attribute 

    '''Instance Attributes (unique to each object) are declared inside the constructor'''

s1 = Student("Ali", 20)
s2 = Student("Muhammad Abu Bakar", 23)

print(s1.name)
print(s1.age)

print("\n--------------------\n")

print(s2.name)
print(s2.age)

#can we update the class attribute?

# for the entire class 

print(s1.school)
print(s2.school)

Student.school = 'NUST'

print(s1.school)
print(s2.school)

# for just one instance of the class / one object 

s1.school = 'LUMS'

print(s1.school)
print(s2.school)

'''IMPORTANT TO UNDERSTAND THE UNDERLYING CONCEPT HERE/WHAT HAPPPENS UNDERNEATH THE HOOD
When we do s1.school = 'LUMS'
Pyhton doesn't modify the class attribute 
what it does it it creates a new instance attribute on s1 which shadows the class attribute 

'''
```
