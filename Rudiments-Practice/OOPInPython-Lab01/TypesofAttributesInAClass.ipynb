```python
'''
In Python attributes are variables that belong to a class, and are used to store data related to an object.

They are of 2 main types
1- Instance Attributes 
    Defined inside th __init__() constructor using self
    each object has its own copy of these attributes 
'''

class Student:
    school = 'Bahria College Karachi NORE-1'    #class attribute

s1 = Student()
s2 = Student()

print(s1.school)
print(s2.school)
```

```python
'''
2- Class Attributes 
    Defined directly inside the class outside any method 
    shared by all the instances of the class 

'''

class Student:

    def __init__(self, name, age):
        self.name = name    #instance attribute 
        self.age = age      #instance attribute 

s1 = Student('Muhammad Abu Bakar', 19)
s2 = Student('Muhammad Ali', 16)

print(s1.name, s1.age)
print(s2.name, s2.age)

```
