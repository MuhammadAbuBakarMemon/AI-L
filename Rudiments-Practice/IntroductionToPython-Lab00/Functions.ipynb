```python
#Non-Parametric Functions 

#the function's definition 
def non_paraFunc():
    print("This is a Non Parametric Function")

non_paraFunc()
```

```python
def paraFunc(a, b):
    """This is a non parametric function"""
    print(a + b)

# takes and printss teh value of a + b

paraFunc(7, 2)
```

```python
# A Function with a return statement 

def sum(arg1, arg2):
    t = arg1 + arg2
    print("Inside the Function: ", t)

    return t

total = sum(2, 23)

print("Outside the Function, courtesy of the return statement: ", total)
```

```python
# A Function without a return statement 

def print_message(message):
    print('Print Statement: ', message)
    return # this line is equivalent to return None

result = print_message("Assalam u Alaikum, Duniya!!!")

#If we try and print the result, we get None, because nothing was retunrned
print(result)
```
