```python
# The For Loop

for var in range(5):
    print(var)
```

```python
# An interesting way to iterate through a string literal 

for letter in 'python':

    print("Current Letter: ", letter)
```

```python
# traversal through a set, the output would be unique everytime as a set is unordered in python (don't preserve the insertion order)

fruits = {'banana', 'apple', 'mango'}

for fruit in fruits:
    print("Current Fruit: ", fruit)

#In case we want a guaranteed order shifting to a list or sort the set 

print("\n----------------------\n")

# sort the set 

for fruit in sorted(fruits):
    print("Current Fruit: ", fruit)
```

```python
# if we wanted to use a list, this would indeed preserve the order of insertion as we see

fruits = ['banana', 'apple', 'mango']

for fruit in fruits:
    print("Current Fruit: ", fruit)
```

```python
# The While Loop

count = 0

while count < 9:
    print("The Count is: ", count)
    count += 1
```
