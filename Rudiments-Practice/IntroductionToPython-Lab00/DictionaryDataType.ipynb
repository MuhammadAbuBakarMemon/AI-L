```python
# A dictionary contains a key and a value that corresponds to that key, 
# key = index 
# value = data at that index location 

dict = {}

dict['col1'] = "This is column 1"

dict[2] = "This is column 2"

print("=== Created Dictionary ===", dict)

print(dict['col1'])

print(dict[2])
```

```python
tinydict = {'name' : 'Muhammad Abu Bakar', 'code':9898, 'dept' : 'Offfensive Security'}

print(tinydict)

print(tinydict.keys())

print(tinydict.values())
```
