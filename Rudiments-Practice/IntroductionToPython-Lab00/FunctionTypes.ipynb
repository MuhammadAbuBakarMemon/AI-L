```python
"""
Function Types Based on Arguments 

1- Positional 
2- Default 
3- Keyword

"""
```

```python

# 1- Positional: values are assigned by position no check on the datatype,
#  30 goes to name and "Abu Bakar" goes to age
def details(name, age):
    print("Name: ", name)
    print("Age: ", age)

# the order in which the arguments are passed do matter 
details(30, "Abu Bakar")
```

```python
# 2- Default: the age = 13 is used when no value is passed 
# Default parameters must come after the non dafualt  parameters in the functions definition

def details(name, age = 13):
    print("Name: ", name)
    print("Age: ", age)

details("Muhammad Abu Bakar")
```

```python
# 3 - Keyword: the order doesn;t matter, when calling the function the value is assigned by the variable's name, so you can pass them in any order 

def details(name, age, height):
    print("Name: ", name)
    print("Age: ", age)
    print("Height: ", height)

details(name = 'Muhammmad Abu Bakar', height = 6.1, age = 30)
```

```python
'''
Functions Types Based on Parameters

1- Varianle Length Parameters
2- Fixed Length Parameters 

'''
```

```python
#1- Fixed Length Prameters 

def add(a, b):
    '''A and B are fixed parameters'''
    return a + b

#providing extra or lesser number of arguments lead to type error
add(1)

add(1, 2, 3)
```

```python
#2- variable Length Parameters/Aruments 

def details(*numbers):
    print('Numbers:', numbers)

details(1, 2, 3)
```

```python
# Keyword variable Length Arguments 

def details(**name):
    print('Name:', name)

details(n1 = 'Ali', n2 = 'Abu Bakar', n3 = 'Fatima')
```

```python
'''
The actual magic here is done courtesy of the * and the **

* a bag that holds items 
** a bag that holds labelled items 

Syntax      Collects            Stored As

*args       positional values  tuple
**kwargs    key=value pairs     dictionary 

The * collects any number of positional arguments into a tuple, w/o it we would have to define exactly how many parameters the function exactly takes 
The ** collects any number of keyword arguments into a dictionary
'''
```
