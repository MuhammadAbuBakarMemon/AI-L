```python
from queue import PriorityQueue

def best_first_search(graph, start, goal):

    visited = set()

    pq = PriorityQueue()
    pq.put((0, start))

    while not pq.empty():

        cost, node = pq.get()

        if node not in visited:
            visited.add(node)
            print(node, end=" ")

        if node == goal:
            print("Goal Reaached.\n")
            return True
        
        for neighbour, weight in graph.get(node, []):
            if neighbour not in visited:
                pq.put((weight, neighbour))

    print("Goal not found in the graph\n")
    return False     

graph = {
    'A': [('B', 3), ('C', 1)], # C has a lower heuristic, so it will be explored first
    'B': [('D', 5), ('E', 4)],
    'C': [('F', 2), ('G', 6)],
    'F': [('I', 0)],           # I is the goal
    'G': [],
    'D': [],
    'E': []
}

print("Best-First Search Path:")
best_first_search(graph, 'A', 'I')
```
