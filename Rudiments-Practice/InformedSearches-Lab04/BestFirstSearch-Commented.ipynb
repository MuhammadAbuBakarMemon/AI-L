```python
from queue import PriorityQueue

def best_first_search(graph, start, goal):
    visited = set()
    pq = PriorityQueue()
    pq.put((0, start)) # Priority queue with priority as the heuristic value
    
    while not pq.empty():
        cost, node = pq.get()
        
        if node not in visited:
            print(node, end=' ')
            visited.add(node)
            
            if node == goal:
                print("\nGoal reached!")
                return True
            
            # Use .get() to avoid KeyErrors on dead-end nodes
            for neighbor, weight in graph.get(node, []):
                if neighbor not in visited:
                    pq.put((weight, neighbor))
                    
    # This must be strictly outside the while loop
    print("\nGoal not reachable!")
    return False

# Example usage with a defined graph
# Format: 'Node': [('Neighbor', Heuristic_Value)]
graph = {
    'A': [('B', 3), ('C', 1)], # C has a lower heuristic, so it will be explored first
    'B': [('D', 5), ('E', 4)],
    'C': [('F', 2), ('G', 6)],
    'F': [('I', 0)],           # I is the goal
    'G': [],
    'D': [],
    'E': []
}

print("Best-First Search Path:")
best_first_search(graph, 'A', 'I')
```
