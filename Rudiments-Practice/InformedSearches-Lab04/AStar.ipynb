```python
graph = {
    'A': {'B': 4, 'C': 3},
    'B': {'E': 12, 'F': 5},
    'C': {'D': 7, 'E': 10},
    'D': {'E': 2},
    'E': {'G': 5},
    'F': {'G': 16},
    'G': {},
}

# Heuristic values for A* search
heuristic = {
    'A': 14, 
    'B': 12, 
    'C': 11, 
    'D': 6, 
    'E': 4, 
    'F': 11, 
    'G': 0 
}

def astar(graph, start, goal):

    frontier = [(start, 0 + heuristic[start])]
    visited = set()
    g_cost = {start : 0}
    came_from = {start : None}

    while frontier:

        frontier.sort(key=lambda x : x[1])

        c_node, current_f = frontier.pop(0)

        if c_node in visited:
            continue 

        visited.add(c_node)
        print(c_node, end=" ")

        if c_node == goal:
            path = []
            while c_node is not None:
                path.append(c_node)
                c_node = came_from[c_node]

            path.reverse()
            print(f"Goal is found with A* path: {path}.\n")
            return 
        
        for neighbour, cost in graph.get(c_node, {}).items():

            new_g_cost = g_cost[c_node] + cost
            f_cost = new_g_cost + heuristic[neighbour]

            
            if neighbour not in g_cost or new_g_cost < g_cost[neighbour]:
                g_cost[neighbour] = new_g_cost
                came_from[neighbour] = c_node
                frontier.append((neighbour, f_cost))

    return "Goal not found"


print("Run A* Search")
astar(graph, 'A', 'G')
```
